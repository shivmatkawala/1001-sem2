### Conditional Statements

```python
# Write a program to provide the meaning idividuals favourate 
# primary color.
# Red, Blue, Green

# NOTE: Always Define a first condition with "if" keyword
# NOTE: Rest of the conditions must be defined with "elif" or "else if"
# NOTE: Default Condition always will be written using "else"

fav_prim_color = input("Enter your favourate primary color: ")

if fav_prim_color == "Red":
    print("Sacrifice")
elif fav_prim_color == "Blue":
    print("Peace")
elif fav_prim_color == "Green":
    print("Nature")
else:
    print("Invalid Color")
    
```

```python
# Nested Conditions Example:

# Ask user to enter a favourate number:
    # Number => if
        # if =>number +ve
            # if => number Int
                    # if => number even
                    # else => number odd

            # else => number Float
        # else=> number -ve
            # if => number Int
                    # if => number even
                    # else => number odd

            # else => number Float

    # Non-Number => else


# Write a program to print what type number is entered by user.
try:
    fav_num = eval(input("Enter your favourate number: "))
except NameError:
    print("Enter numeric values")
else:
    if isinstance(fav_num, int):
        print(f"{fav_num} is Integer")
        if fav_num > 0:
            print(f"{fav_num} is +ve")
            if fav_num % 2 == 0:
                print("Even")
            else:
                print("Odd")
        elif fav_num == 0:
            print(f"{fav_num} is Neutral")
            print(f"{fav_num} is Even")
        

        else:
            print(f"{fav_num} is -ve")
            if fav_num % 2 == 0:
                print(f"{fav_num} is Even")
            else:
                print(f"{fav_num} is Odd")
    else:
        print(f"{fav_num} is Float")
        if fav_num > 0:
            print(f"{fav_num} is +ve")
        else:
            print(f"{fav_num} is -ve")
```

### Match Case

```python
# It is pythons pattern matching feature introduced in 3.10

# 0, 1, 2, 3, 4, 5, 6

day_num = int(input("Enter a day number: "))

match day_num:
    case 0:
        print("Sunday")
    case 1:
        print("Munday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case _:
        print("Invalid day number.")
```

```python
command = input("Enter command: ")

match command:
    case "start":
        print("Vehicle Started")
    case "stop":
        print("Vehicle Stopped")
    case "reverse":
        print("Vehicle Reversed")
    case _:
        print("Unknown Command")
```

```python
point = (3, 9)

match point:
    case (0, 0):
        print("Origin")
    case (x, 0):
        print(f"X-axis: {x}")
    case (0, y):
        print(f"Y-axis: {y}")
    case(x, y):
        print(f"Point: {x}, {y}")
```
