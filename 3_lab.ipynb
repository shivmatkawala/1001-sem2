Write a program to find all Armstrong numbers in
between 100 and 10000.


153 => 1**3 + 5**3 + 3**3
     => 1+125+27
     => 153

```python
for num in range(100, 10000):
    power = len(str(num))
    total = 0
    for digit in str(num):
        total+=(int(digit)**power)
    if total == num:
        print(num)
```

Write a program to get the result "Pass" or "Fail"

```python
Physics = eval(input("Physics Marks: "))
Chemistry = eval(input("Chemistry Marks: "))
Math = eval(input("Math Marks: "))

percentage = ((Physics+Chemistry+Math)/300)*100

result = "Pass" if percentage >= 50 else "Fail"
print(result)
```

```python
Physics = eval(input("Physics Marks: "))
Chemistry = eval(input("Chemistry Marks: "))
Math = eval(input("Math Marks: "))

percentage = ((Physics+Chemistry+Math)/300)*100

grade = "A+" if percentage >= 90 else "A" if percentage >= 80 else "B" if percentage >= 60 else "C" if percentage >= 50 else "Fail"
print(grade)
```
